```python
# १. Negative Total Amount make (Positive) .
df['total_amount_inr'] = df['total_amount_inr'].abs()

# २. १८% GST re-calculate total amount correction
df['gst_amount_inr'] = (df['base_amount_inr'] * 0.18).round(2)
df['total_amount_inr'] = df['base_amount_inr'] + df['gst_amount_inr']

# ३. check result
df.head()
```

```python
# Step 1: Load all required CSV files with proper path
customers = pd.read_csv(r"C:\Users\Bhagyashri\Downloads\ec4aeca2a8032be0347eb12ca3d29c96\customers.csv")
tickets = pd.read_csv(r"C:\Users\Bhagyashri\Downloads\ec4aeca2a8032be0347eb12ca3d29c96\support_tickets.csv")

# Step 2: Clean customers dataset
customers['acquisition_date'] = pd.to_datetime(customers['acquisition_date'], dayfirst=True, errors='coerce')
customers['date_of_birth'] = pd.to_datetime(customers['date_of_birth'], dayfirst=True, errors='coerce')
customers['city_name'] = customers['city_name'].astype(str).str.strip().str.title()

# Step 3: Calculate total revenue per customer
customer_revenue = df.groupby('customer_id')['total_amount_inr'].sum().reset_index()
customer_revenue.rename(columns={'total_amount_inr': 'total_revenue_inr'}, inplace=True)

# Step 4: Calculate total support tickets per customer
customer_tickets = tickets.groupby('customer_id')['ticket_id'].count().reset_index()
customer_tickets.rename(columns={'ticket_id': 'total_tickets_raised'}, inplace=True)

# Step 5: Merge revenue and tickets data with customers dataset
master_df = pd.merge(customers, customer_revenue, on='customer_id', how='left')
master_df = pd.merge(master_df, customer_tickets, on='customer_id', how='left')

# Step 6: Fill missing values with 0
master_df['total_tickets_raised'] = master_df['total_tickets_raised'].fillna(0)
master_df['total_revenue_inr'] = master_df['total_revenue_inr'].fillna(0)

# Step 7: Save cleaned master dataset to CSV file
master_df.to_csv('NexaTel_Cleaned_Master.csv', index=False)

print("Phase 1 completed successfully! Master Dataset is ready.")
```

### **Phase 2: Feature Engineering & Advanced KPI Calculations**

```python
# Step 1: Fix Warning by adding dayfirst=True
master_df['acquisition_date'] = pd.to_datetime(master_df['acquisition_date'], dayfirst=True, errors='coerce')
master_df['churn_date'] = pd.to_datetime(master_df['churn_date'], dayfirst=True, errors='coerce')

# Step 2: Handle missing acquisition_date values by filling with default/median tenure
today = pd.to_datetime('2026-09-30')
master_df['end_or_today'] = master_df['churn_date'].fillna(today)

# Step 3: Recalculate Tenure and ARPU safely
master_df['tenure_months'] = ((master_df['end_or_today'] - master_df['acquisition_date']).dt.days / 30.44).round(1)
master_df['tenure_months'] = master_df['tenure_months'].fillna(1) # Default to 1 month if missing

master_df['arpu'] = (master_df['total_revenue_inr'] / master_df['tenure_months'].replace(0, 1)).round(2)

# Step 4: Drop temporary helper column
if 'end_or_today' in master_df.columns:
    master_df.drop(columns=['end_or_today'], inplace=True)

# Step 5: Check output again
master_df[['customer_id', 'is_churned', 'tenure_months', 'arpu', 'total_revenue_inr']].head()
```

```python
# Step 1: Save updated feature engineered master dataset
master_df.to_csv('NexaTel_Master_Phase2.csv', index=False)

print("Phase 2 Feature Engineering Completed Successfully!")
```

```python
# Step 1: Calculate Overall Churn Rate & Average ARPU
total_customers = len(master_df)
churned_customers = master_df['is_churned'].sum()
churn_rate = (churned_customers / total_customers) * 100
avg_arpu = master_df['arpu'].mean()

print(f"Total Customers: {total_customers}")
print(f"Total Churned Customers: {churned_customers}")
print(f"Overall Churn Rate: {churn_rate:.2f}%")
print(f"Average ARPU: ₹{avg_arpu:.2f}")
```

### **Phase 3: Exploratory Data Analysis (EDA)**

```python
# Step 1: Churn Rate by City (Top 10 Cities with highest churn)
city_churn = master_df.groupby('city_name').agg(
    total_customers=('customer_id', 'count'),
    churned_customers=('is_churned', 'sum')
).reset_index()

city_churn['churn_rate_%'] = ((city_churn['churned_customers'] / city_churn['total_customers']) * 100).round(2)
top_churn_cities = city_churn.sort_values(by='churn_rate_%', ascending=False).head(10)

print("--- Top 10 High Churn Cities ---")
print(top_churn_cities)

# Step 2: Impact of Support Tickets on Churn
ticket_churn = master_df.groupby('total_tickets_raised').agg(
    total_customers=('customer_id', 'count'),
    churned_customers=('is_churned', 'sum')
).reset_index()

ticket_churn['churn_rate_%'] = ((ticket_churn['churned_customers'] / ticket_churn['total_customers']) * 100).round(2)

print("\n--- Churn Rate by Support Tickets Raised ---")
print(ticket_churn.head(10))
```

```python
# Step 1: Churn Rate for major cities (having at least 100 customers)
major_city_churn = city_churn[city_churn['total_customers'] >= 100].sort_values(by='churn_rate_%', ascending=False)

print("--- Major Cities High Churn Rate (Minimum 100 Customers) ---")
print(major_city_churn.head(10))

# Step 2: Save final EDA dataset for Power BI visualization
master_df.to_csv('NexaTel_Master_EDA_Final.csv', index=False)

print("\nPhase 3 EDA Completed Successfully! Dataset ready for Power BI.")
```
